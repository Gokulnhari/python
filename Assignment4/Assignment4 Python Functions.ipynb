3. Write a Python function find_maximum(numbers) that takes a list of integers and returns the maximum value without using the built-in max() function. Use a loop to iterate through the list and compare values.


```python
def find_maximum(numbers):
    if not numbers:
        return None
    max_value = numbers[0]
    for num in numbers:
        if num > max_value:
            max_value = num
    return max_value

print("The maximum value is:", find_maximum([3, 5, 7, 2, 8]))
```

4. Explain the difference between local and global variables in a Python function. Write a program where a global variable and a local variable have the same name and show how Python differentiates between them.


Global variables: Defined outside any function. They can be accessed and modified within a function using the global keyword.
Local variables: Defined inside a function. Their scope is limited to that function — they can’t be accessed outside it.


```python
x = 10   # This is a global variable

def example_function():
    x = 5  # This is a local variable
    print("Local x:", x)

example_function()
print("Global x:", x)
```

5. Create a function calculate_area(length, width=5) that calculates the area of a rectangle.
If only the length is provided, the function should assume the width is 5.

```python
def calculate_area(length, width=5):
    area = length * width
    return area

# Calling the function with both arguments.
print("Area with length 10 and width 3:", calculate_area(10, 3))
# Calling the function with only the length argument.
print("Area with length 10:", calculate_area(10))
```
